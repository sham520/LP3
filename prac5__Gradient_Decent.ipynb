```python
def f(x):
    return(x+3)**2
```

```python
def derivate(x):
    return 2*(x+3)
```

```python
x =2
learning_rate =0.1
iterations =50
```

```python
for i in range(iterations):
    x=x-learning_rate*derivate(x)
```

```python
print("The Loacl minima is:",round(x,6))
print("the Minimun value of y is:",round(f(x),6))
```
